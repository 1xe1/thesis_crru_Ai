```python
# นำเข้า YOLO จาก ultralytics
from ultralytics import YOLO
```

```python
# กำหนดโมเดลที่ต้องการฝึก (เลือกจาก yolov8n, yolov8s, yolov8m, yolov8l, หรือ yolov8x)
model = YOLO('licensPlate.pt')  # โมเดลขนาด small
```

```python
# ทำการ fine-tuning model ด้วย dataset ใหม่
results = model.train(
    data='datasets/data.yaml',  # ระบุไฟล์ data.yaml ที่กำหนด dataset ใหม่
    epochs=50,                         # จำนวน epochs ที่จะเทรน
    batch=16,                          # ขนาด batch size
    imgsz=640,                         # ขนาดภาพที่ใช้ในการเทรน (เช่น 640x640)
    lr0=0.01,                          # ค่าความเร็วในการเรียนรู้ (learning rate)
    pretrained=True,                   # ใช้ model pre-trained
    patience=5,                        # Early stopping: หยุดเทรนถ้าผลลัพธ์ไม่ดีขึ้นใน 5 epochs
    augment=True,                      # ใช้ data augmentation เช่นการหมุน/สะท้อนภาพ เพื่อลด overfitting
    val=True,                          # ตรวจสอบ validation set ระหว่างเทรนเพื่อป้องกัน overfitting
    project='runs/pretrain',           # กำหนด path สำหรับบันทึกผลลัพธ์
    name='newLicensePlate',                  # ชื่อโปรเจกต์หรือชื่อการเทรน
    optimizer='Adam',                  # เลือก optimizer เช่น Adam หรือ SGD
    workers=4,                         # จำนวน workers สำหรับการโหลดข้อมูล
)
```
