```python
# นำเข้า YOLO จาก ultralytics
from ultralytics import YOLO
```

```python
# กำหนดโมเดลที่ต้องการฝึก (เลือกจาก yolov8n, yolov8s, yolov8m, yolov8l, หรือ yolov8x)
model = YOLO('helmet.pt')  # โมเดลขนาด small
```

```python
# ฝึกโมเดล
results = model.train(
    data='Helmetdatasets/data.yaml',
    epochs=50,
    imgsz=640,
    batch=8,  # ลด batch size เช่น จาก 16 เป็น 8 หรือ 4
    device=0,
    plots=True
)
```
